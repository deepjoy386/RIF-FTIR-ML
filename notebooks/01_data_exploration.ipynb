```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Display plots inside the notebook
%matplotlib inline

# Better looking plots
plt.style.use("ggplot")

# Show all columns if needed
pd.set_option("display.max_columns", None)
```

```python
data = pd.read_csv("../data/RIF_FTIR.csv")

print("Shape:", data.shape)

data.head()
```

```python
# ----------------------------
# Convert FTIR data to ML format
# ----------------------------

# Extract labels from the first row
labels = data.iloc[0, 1:]

# Extract wavenumbers (skip the first "group" row)
wavenumbers = data.iloc[1:, 0].astype(float)

# Extract spectral values
spectra = data.iloc[1:, 1:]

# Make wavenumbers the column names
spectra.index = wavenumbers

# Transpose so rows = patients
spectra = spectra.T

# Convert everything to float
spectra = spectra.astype(float)

# Add labels
spectra["Label"] = labels.values

spectra.head()
```

```python
print("Shape:", spectra.shape)

spectra.info()
```

```python
spectra["Label"] = spectra["Label"].map({
    "control": 0,
    "RIF": 1
})

spectra.head()
```

```python
spectra["Label"].value_counts()
```

```python
X = spectra.drop(columns=["Label"])
y = spectra["Label"]

print(X.shape)
print(y.shape)
```

```python
wavenumbers = X.columns.astype(float)

print(wavenumbers[:10])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.style.use("default")

for _, row in spectra.iterrows():
    plt.plot(
        wavenumbers,
        row[:-1],
        linewidth=0.6,
        alpha=0.25,
        color="steelblue"
    )

# FTIR convention
plt.gca().invert_xaxis()

plt.xlabel("Wavenumber (cm$^{-1}$)", fontsize=13)
plt.ylabel("Absorbance (a.u.)", fontsize=13)

plt.title(
    "Raw FTIR Spectra of Control and RIF Samples",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.grid(False)

plt.tight_layout()

plt.savefig(
    "../figures/Figure1_RawSpectra.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Process before calculating group means (smoothing + vector normalization)
import sys
from pathlib import Path
sys.path.append(str(Path("../src").resolve()))
from preprocess import smooth_spectra, vector_normalize

X_processed = vector_normalize(smooth_spectra(X))
processed_spectra = X_processed.copy()
processed_spectra["Label"] = y.values

# Separate processed spectra by group
control = processed_spectra[processed_spectra["Label"] == 0].drop(columns=["Label"])
rif = processed_spectra[processed_spectra["Label"] == 1].drop(columns=["Label"])

# Mean spectra
control_mean = control.mean(axis=0)
rif_mean = rif.mean(axis=0)

# Standard deviation
control_std = control.std(axis=0)
rif_std = rif.std(axis=0)

# Convert column names (wavenumbers) to float
wn = control.columns.astype(float)
```

```python
import matplotlib.pyplot as plt

plt.style.use("default")

plt.figure(figsize=(10,6))

# Control
plt.plot(
    wn,
    control_mean,
    label="Control",
    linewidth=2.2,
)

plt.fill_between(
    wn,
    control_mean - control_std,
    control_mean + control_std,
    alpha=0.15,
)

# RIF
plt.plot(
    wn,
    rif_mean,
    label="RIF",
    linewidth=2.2,
)

plt.fill_between(
    wn,
    rif_mean - rif_std,
    rif_mean + rif_std,
    alpha=0.15,
)

# FTIR convention
plt.gca().invert_xaxis()

plt.xlabel("Wavenumber (cm$^{-1}$)", fontsize=13)
plt.ylabel("Absorbance (a.u.)", fontsize=13)

plt.title(
    "Mean Processed FTIR Spectra of Control and RIF Samples",
    fontsize=15,
    fontweight="bold"
)

plt.legend(fontsize=11)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.grid(False)

plt.tight_layout()

plt.savefig(
    "../figures/Figure2_MeanSpectra.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
