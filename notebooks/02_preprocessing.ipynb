```python
import sys
from pathlib import Path

sys.path.append(str(Path("../src").resolve()))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from preprocess import (
    smooth_spectra,
    vector_normalize
)
```

```python
spectra = pd.read_csv(
    "../data/processed_spectra.csv",
    index_col=0
)

X = spectra.drop(columns=["Label"])
y = spectra["Label"]
```

```python
X_smooth = smooth_spectra(X)

X_processed = vector_normalize(X_smooth)
```

```python
sample = 0

wavenumbers = X.columns.astype(float)

plt.figure(figsize=(12,6))

plt.plot(
    wavenumbers,
    X.iloc[sample],
    label="Original",
    linewidth=2
)

plt.plot(
    wavenumbers,
    X_smooth.iloc[sample],
    label="Smoothed",
    linewidth=2
)

plt.gca().invert_xaxis()

plt.xlabel("Wavenumber (cm$^{-1}$)")
plt.ylabel("Absorbance")

plt.title("Effect of Savitzky-Golay Smoothing")

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    wavenumbers,
    X_smooth.iloc[sample],
    label="Smoothed"
)

plt.plot(
    wavenumbers,
    X_processed.iloc[sample],
    label="Normalized"
)

plt.gca().invert_xaxis()

plt.legend()

plt.title("Effect of Vector Normalization")

plt.show()
```

```python
plt.tight_layout()

plt.savefig(
    "../figures/figure_01_smoothing.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
